```python
# import sys
# import argparse
import json
from urllib import request
from urllib.parse import quote, urlencode
```

```python
import xml.etree.ElementTree as ET
```

```python
url = "https://home.treasury.gov/resource-center/data-chart-center/interest-rates/pages/xml?data=daily_treasury_yield_curve&field_tdr_date_value_month=202609"
```

```python
req = request.Request(url)
with request.urlopen(req) as conn:
    # tree = ET.fromstring(conn.read().decode("utf-8"))
    tree = ET.parse(conn)
```

```python
with request.urlopen(req) as conn:
    namespaces = {
        prefix: uri 
        for event, (prefix, uri) in ET.iterparse(conn, events=['start-ns'])
    }
namespaces
```

```python
root = tree.getroot()

# # Option B: Parse from a string constant
# xml_string = '<data><item id="1">Python</item></data>'
# root = ET.fromstring(xml_string)

# # Verify the root tag
# print(root.tag)  # Output: data
```

```python
for entry in root.findall('entry', namespaces):
    print(entry)
    for props in entry.findall('m:properties'):
        print(props)
```

```python
from datetime import datetime
```

```python
ns = {'atom': 'http://www.w3.org/2005/Atom',}

for entry in root.findall('atom:entry', ns):
    # name = actor.find('real_person:name', ns)
    print(entry)
    for content in entry.findall('atom:content', ns):
        print(content)
        for props in content:
            recd = { "feed_id": -1, "date": None }
            for prop in props:
                proptag_ns = namespaces['d']
                proptag_ns_len = len(proptag_ns) + 2 # + 2 for curly braces
                print(f"{prop.tag[proptag_ns_len:]}: {prop.text}")
                proptag_adj = prop.tag[proptag_ns_len:]
                if (proptag_adj == "Id"):
                    recd["feed_id"] = int(prop.text)
                elif proptag_adj == "NEW_DATE":
                    recd["date"] = datetime.fromisoformat(prop.text)
                else:
                    recd[proptag_adj] = float(prop.text)
            print(recd)
```
